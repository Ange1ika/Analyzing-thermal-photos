```python
x = 110
y = 10
pixel_flame_gray8 = img[y, x]
```

```python
def pixel_to_temperature(pixel_val, t_min, t_max, max_pixel_val=255):
    """
    Преобразует значение пикселя в температуру на основе мин/макс температуры.
    
    Args:
        pixel_val (int or float): Значение пикселя (0–255 или 0–65535).
        t_min (float): Минимальная температура в кадре (°C).
        t_max (float): Максимальная температура в кадре (°C).
        max_pixel_val (int): Максимальное значение пикселя (по умолчанию 255).
    
    Returns:
        float: Температура в °C.
    """
    return t_min + (pixel_val / max_pixel_val) * (t_max - t_min)
```

```python
# write pointer
pixel_flame_gray16 = pixel_to_temperature(pixel_flame_gray8, 3.93, 21.36)
cv2.circle(img, (x, y), 2, (0, 0, 0), -1)
print(pixel_flame_gray16)
#cv2.circle(gray16_image, (x, y), 2, (0, 0, 0), -1)
# write temperature value in gray8 and gray16 image
cv2.putText(img,"{0:.1f} Celc".format(pixel_flame_gray16),(x - 80, y - 15), cv2.FONT_HERSHEY_PLAIN, 1,(255,0,0),2)
#cv2.putText(gray16_image,"{0:.1f} Fahrenheit".format(pixel_flame_gray16),(x - 80, y - 15), cv2.FONT_HERSHEY_PLAIN, 1,(255,0,0),2)
 
plt.figure(figsize=(8, 6))
plt.imshow(img)
plt.title('calculate temper')
plt.axis('off')
plt.show()
```
