```python
import cv2
import matplotlib.pyplot as plt
%matplotlib
```

```python
def get_maps(img, ksize=11):

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    src = cv2.GaussianBlur(gray, (ksize, ksize), 0)
    laplacian = cv2.Laplacian(src, cv2.CV_64F)
    sharpness_map = cv2.convertScaleAbs(laplacian)
    img_name = "/home/angelika/Desktop/Aramco/test_projects/bookmarks"

    plt.imshow(sharpness_map)
    plt.colorbar(orientation='vertical', fraction=.02)
    plt.axis('off')
    plt.savefig(f'{img_name}_laplacian_abs.png', format='png', dpi=300)
    plt.show()
    
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    laplacian = cv2.Laplacian(gray, cv2.CV_64F)
    sharpness_map = laplacian.astype('uint8')
    
    plt.imshow(sharpness_map)
    plt.colorbar(orientation='vertical', fraction=.02)
    plt.axis('off')
    plt.savefig(f'{img_name}_laplacian.png', format='png', dpi=300)
    plt.show()
```

```python
img = cv2.imread("/home/angelika/Desktop/Aramco/test_projects/bookmarks.jpg")
get_maps(img)
```
